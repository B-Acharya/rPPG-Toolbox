```python
def plot_model_metrics(model_metrics):
    # Plot MAE and Variance for different models
    models = list(model_metrics.keys())
    mae_values = [model_metrics[m]['MAE'] for m in models]
    var_values = [model_metrics[m]['VAR'] for m in models]

    plt.figure(figsize=(10, 5))
    plt.bar(models, mae_values, yerr=var_values, capsize=5, color='skyblue', label='MAE with Variance')
    plt.ylabel('MAE')
    plt.xlabel('Models')
    plt.title('MAE and Variance Across Different Models')
    plt.legend()
    plt.show()
```
